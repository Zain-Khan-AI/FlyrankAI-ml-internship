## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd
import numpy as np

# Load the exact dataset path
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Metric 1: Total unique clients and data size
total_rows = len(df)
unique_clients = df['client_id'].nunique()

# Metric 2: Identify at-risk rows using the 'trend_direction' column
downward_rows = len(df[df['trend_direction'] == 'DOWN'])
pct_downward = (downward_rows / total_rows) * 100

print(f"Total rows analyzed: {total_rows:,}")
print(f"Total unique client accounts mapped: {unique_clients:,}")
print(f"Content rows showing active organic downward trend: {downward_rows:,} ({pct_downward:.2f}%)")
```
