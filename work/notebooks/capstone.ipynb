## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

* **Core Assumptions:** We assume that content size characteristics (word_count) demonstrate highly correlated structural interactions with search engine exposure limits (impressions) and conversion success (clicks, ctr, position), and that these latent organic behavior modes form distinct geometric clusters.
* **Model Input Vector Matrix:** The final vector mapping utilizes 5 scaled spatial features: log_impressions, log_clicks, avg_position, ctr, and log_word_count.
* **Mathematical Transformations:** Saturated log transformations (np.log1p) were implemented across right-skewed scale attributes (impressions, clicks, word count) to stabilize variance vectors. Standard feature scaling normalization via Z-scores (Mean=0, Std=1) was applied to reconcile diverse metric domains.
* **Baseline Standard:** A hard-coded, single-dimension heuristic baseline stratifying content assets strictly by impression volume quartiles.
* **Validation Split Architecture & Data Leakage Protections:** A rigorous time-aware window isolation split design was strictly enforced. Standard scaler scaling profiles and K-Means geometric centroid parameters were calculated and locked entirely using the February horizon data frame (fit_transform). The parameters were subsequently applied to March matrices strictly inside a downstream operational stream using frozen transformations (.transform(), .predict()), successfully preventing any future informational leakage.

## 5. Limitations

*What this work cannot claim.*

This research provides critical decision-support context but operates under strict structural boundaries. **This model does not establish explicit causal fresh-refresh impact parameters or structural ranking factors within Google's algorithmic components.** Observed feature configurations represent strictly non-causal spatial associations. Furthermore, results are fundamentally bounded by the data quality boundaries of the source archive: a severe metadata gap was measured where 75% of active February content nodes lacked valid structural parameters in dim_content, meaning the cluster allocations only generalize safely to established, structurally documented URL architectures.


## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

Based on the validated time-aware cluster tracking behaviors, the analytical playbooks are ranked by projected operational efficiency:

1. **Priority 1: Protect & Defend (Superstars):** These high-performance items transform low exposure metrics into efficient organic traffic dividends. Strategy: Freeze technical structures, guard against code regressions, and prevent content alterations.

4. **Priority 4: Strategic Pruning Consolidation (Zombies):** Massive volumes of zero-yield pages drag down enterprise site crawling efficiency boundaries. Strategy: Systematically consolidate high-overlap thematic groupings into centralized source documents, and completely prune dead URLs to optimize crawl budgets.
