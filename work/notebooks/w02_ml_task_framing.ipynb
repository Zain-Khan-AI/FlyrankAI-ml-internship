## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

##### My lane as an ML task: Classification

My ML task is **classification**. The goal is to predict whether a content item is declining or not declining based on its available performance and content characteristics.

This is a classification problem because the prediction has two possible outcomes: declining or not declining. The model would classify each content item into one of these two categories.

I chose this framing because identifying declining content can help a content team decide which pages should be prioritized for content refresh and optimization.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


I would predict whether a content item is declining or not declining.

Because the starter dataset does not contain an `is_declining_label` column, I use a defined proxy called `declining_proxy`. The proxy is created from the observed `trend_direction` field.

The rule is:

- `trend_direction == "down"` → `declining_proxy = 1`
- Otherwise → `declining_proxy = 0`

This is a defined rule based on an observed trend signal, rather than a directly observed future business outcome. The proxy gives the ML task a clear binary target for identifying content that may need a refresh.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I would use **F1-score** as the main success metric.

F1-score balances precision and recall, which is useful for this task because both types of mistakes matter. We want to identify genuinely declining content without incorrectly flagging too much content that is not declining.

A higher F1-score means the model is better at identifying declining content while keeping false alerts under control. I would consider the model useful if it achieves a substantially better F1-score than a simple baseline rule.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd
data_path = "../../data/raw/content_refresh_anonymized.csv"
# Load the starter dataset
df = pd.read_csv(data_path)

# Create the target/proxy
df["declining_proxy"] = (df["trend_direction"] == "down").astype(int)

# Select the relevant columns for this ML task
lane_df = df[
    [
        "content_id",
        "client_id",
        "content_type",
        "search_volume",
        "competition",
        "word_count",
        "ctr",
        "engagement_rate",
        "scroll_rate",
        "avg_position",
        "declining_proxy"
    ]
].copy()

# Show the real dataframe
lane_df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

#### Why ML beats a fixed rule

A simple rule such as "if trend_direction is down, mark the content as declining" can identify the current trend, but it does not use the wider set of signals available in the data.

Content performance can be influenced by multiple factors, including search volume, competition, word count, CTR, engagement rate, scroll rate, and average position. These signals can interact in ways that are difficult to capture with one fixed if-statement.

ML can learn patterns across these different signals and identify combinations that are associated with declining content. This can help the content team prioritize which pages are most likely to need a refresh instead of relying on a single rule.
